# ENVS456: Brent Crude Oil Time-Series Forecasting with AR(12) & Interactive Plotly Dashboard
This notebook fits an Autoregressive `AR(12)` time-series model to monthly Brent Crude prices, evaluates stationarity and out-of-sample error metrics, and exports a standalone interactive HTML dashboard.

```python
import sys
sys.path.append('..')
from src.ar12_model import generate_brent_oil_series, OilPriceAR12Forecaster
from src.dashboard import export_interactive_plotly_dashboard

df = generate_brent_oil_series(n_months=180, seed=42)
forecaster = OilPriceAR12Forecaster(lags=12)
results = forecaster.fit_and_forecast(df, horizon=12, test_size=12)
print('Out-of-Sample Metrics:', results['metrics'])
export_interactive_plotly_dashboard(results, '../output/brent_ar12_dashboard.html')
```
